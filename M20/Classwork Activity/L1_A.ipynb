```python
# Import libaries
from transformers import pipeline
from IPython.display import display, HTML, clear_output

# Load Hugging Face sentiment model (free, no API key)
sentiment_analyzer = pipeline("sentiment-analysis")

# Store history
history = []

# function to display styled sentiment result
def show_result(text, label, score):
  if label == "POSITIVE":
    color = "green"
    emoji = "😊"
  else:
    color = "red"
    emoji = "😠"

  display(HTML(f"""
  <div style="
    border:2px solid {color};
    padding:15px;
    border-radius:10px;
    margin:10px 0;
    font-family:Arial;">

    <h3 style="color:{color};">{emoji} Sentiment: {label}</h3>
    <p><b>Confidence:</b> {round(score,2)}</p>
    <p><b>Your Text:</b> {text}</p>
  </div>
  """))

# Function to show history
def show_history():
  if not history:
    display(HTML("<h4>No history yet!<h4>"))
    return

  html = "<h3> 📜 History</h3>"

  for i, (text, label, score) in enumerate(history, start=1):
    color = "green" if label == "POSITIVE" else "red"
    html += f"""
    <div style="margin:5px; padding:8px; border-left:5px solid {color};">
      {i}. {text} → <b style="color:{color}>{label}</b> ({round(score,2)})
    </div>
    """

  display(HTML(html))

# Function to show stats
def show_stats():
  if not history:
    display(HTML("<h4>No data available!</h4>"))
    return

  positive = sum(1 for _, label, _ in history if label == "POSITIVE")
  negative = sum(1 for _, label, _ in history if label == "NEGATIVE")
  total = len(history)

  display(HTML(f"""
  <h3>📊 Statistics</h3>
  <p>Total Inputs: <b>{total}</b></p>
  <p style="color:green;">Positive: {positive}</p>
  <p style="color:red;">Negative: {negative}</p>
  <p> Positive %: {round((positive/total)*100,2)}%</p>
  <p> Negative %: {round((negative/total)*100,2)}%</p>
  """))

# Main Loop
def run_analyzer():
  print("🎯 Sentiment Analyzer (Colab Version)")
  print("Commands: history | stats | clear | exit\n")

  while True:
    user_input = input("Enter text: ")

    if user_input.lower() == "exit":
      print("Goodbye 👋")
      break

    elif user_input.lower() == "history":
      show_history()

    elif user_input.lower() == "stats":
      show_stats()

    elif user_input.lower() == "clear":
      clear_output()

    else:
      result = sentiment_analyzer(user_input)
      label = result[0]['label']
      score = result[0]['score']

      # Save history
      history.append((user_input, label, score))

      # Show styled result
      show_result(user_input, label, score)

# Run the app
run_analyzer()
```
